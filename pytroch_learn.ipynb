```python
import torch
import matplotlib.pyplot as plt
```

```python
# Temporary weights and biases for linear regression
w = 0.7
b = 0.3
```

```python
# to create a dummy dataset
start = 0 
end = 1
step = 0.02
```

```python
X = torch.arange(start,end,step).unsqueeze(dim=1)
y = w * X * b
```

```python
# Train Test split
train_split = int(0.8 *len(X))
```

```python
X_train = X[:train_split]
y_train = y[:train_split]

X_test = X[train_split:]
y_test = y[train_split:]
```

```python
len(X_train),len(y_train),len(X_test),len(y_test)
```

```python
def plot_predications(train_data = X_train,train_labels = y_train,
                     test_data = X_test,test_labels=y_test, predictions = None, validation_data=None, validation_lables=None):
    plt.figure(figsize=(10,7))
    plt.scatter(train_data, train_labels, c="b", marker="*", s=8, label="Training data")
    plt.scatter(test_data, test_labels, c="g", marker="o", s=8, label="Test data")
    if predictions is not None:
        plt.scatter(test_data, predictions, c="r", s=4, label="predictions")
    if validation_data is not None:
        plt.scatter(validation_data, validation_lables, c="r", s=4, label="Validation data")
    
    plt.legend(prop={"size":11});
    plt.show()
    
    
```

```python
plot_predications()
```

```python
# Build a pytorch based linear regression model

from torch import nn


class LinearRegressionModel(nn.Module):

    def __init__(self):
        super().__init__()
        self.weights = nn.Parameter(torch.randn(1,
                                                requires_grad=True,
                                                dtype=torch.float))
        self.bias = nn.Parameter(torch.randn(1,
                                             requires_grad=True,
                                             dtype=torch.float))

    def forward(self, x: torch.tensor) -> torch.tensor:
        return self.weights * x * self.bias
```
